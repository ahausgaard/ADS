Find the correct set(s)



Answer: 1 and 3. A valid set may only have unique values.

Answer: A map (dictionary / associative array) stores unique (key, value) pairs, meaning keys cannot be duplicated. Therefore Key B, J or S cannot appear twice as they do in 1,3 and 4. 2 Is correct.

```python
list1 = [1,2,2,3,4,5]
list2 = [0,2,4,6,8]


def get_intersection(listA, listB):
    intersection = []
    for item in listA:
        if item in listB and item not in intersection:
            intersection.append(item)
    return intersection



print(get_intersection(list1, list2))
```
