Easy Exercise 1:
Implement a function that finds the sum, average, min and max for the numbers in an
unsorted list
Expand it by finding the frequency of all the numbers

```python
from typing import NamedTuple

list_x = (1,3,5,2,-2,5,7,3,4)

def findStuff(myList):
    sum = 0
    min = float('inf')
    max = -float('inf')
    frequency = []
    for i in myList:
        sum += i
        if i < min:
            min = i
        if i > max:
            max = i
        found = False
        for j in frequency:
            if j[0] == i:
                j[1] += 1
                found = True
                break
        if not found:
            frequency.append([i, 1])

    avg = sum/len(myList)
    print(f"Average:{avg}, Sum:{sum}, Min:{min}, Max:{max}")
    print(f"Frequency:{frequency}")

findStuff(list_x)
```

Easy Exercise 2:
Implement a function that finds the frequency of all numbers in a sorted list.

```python
list_x = (1,3,5,2,-2,5,7,3,4)
list_x_sorted = sorted(list_x)

def findFreq(numbers):
    frequency = []

    for i in numbers:
        if not frequency or frequency[-1][0] != i:
            frequency.append([i, 1])
        else:
            frequency[-1][1] += 1

    return frequency


print(findFreq(list_x_sorted))
```

Exercise 1: What is the sum?
Implement a recursive function that returns the sum of all digits in a positive integer
(e.g. sum_digits(222) → 2+2+2 = 6)
1. What is the base case
2. What is the time complexity?

```python
number = 23069

#Iterativ
def sum_digits(myNumber):
    sum = 0
    while myNumber > 0:
        sum += myNumber % 10
        myNumber = myNumber // 10
    print(sum)

sum_digits(number)

#Rekursiv udgave:
def digitSumRecursive(n):
    if n < 10:
        return n
    return n % 10 + digitSumRecursive(n // 10)

digitSumRecursive(number)
```

Exercise 2: Unique vowels
Implement a recursive function that returns all unique vowels
(e.g. unique_vowels(”Recursion is great!”) → [’e’, ’u’, ’i’, ’o’, ’a’])
1. What is the base case
2. What is the time complexity?

```python
sentence = "Recursion is great!".lower()
VOWELS = ['a', 'e', 'i', 'o', 'u', 'y','æ', 'ø', 'å']

#Iterativ
def unique_vowels(mySentence):
    unique = []
    for letter in mySentence:
        if letter in VOWELS and letter not in unique:
            unique.append(letter)
    print(unique)

unique_vowels(sentence)

#Rekursiv
def unique_vowels_rec(mySentence, i=0, unique=None):
    if unique is None:
        unique = []
    if i == len(mySentence):
        return unique

    letter = mySentence[i]
    if letter in VOWELS and letter not in unique:
        unique.append(letter)

    return unique_vowels_rec(mySentence, i + 1, unique)

unique_vowels_rec(sentence)
```

Exercise 3: Coffee Inheritance
Create a Coffee class with the attributes:
- name
- price

Create two subclasses:
- Espresso
- Latte

Each subclass should:
- Have its own default price
- Have a to_string function for printing its content

```python
#Jeg ignorerede nedarvning.. Synes ikke opgaven gav mening.
class Coffee:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        print(self.name,": ", self.price, "kr.")

espresso = Coffee("Espresso", 42)

latte = Coffee("Latte", 50)

espresso.__str__()
latte.__str__()
```

Exercise 5: Binary search
Implement a function that uses binary search on a sorted list of ints, to find out how many
times an int occurs in the list:
Hint: find the index of left-most occurence and right-most and calculate the distance
between these


```python
number = 2
list=[1,2,3,3,3,4,5,12,213]
def binarySearch(value, myList):
    length = len(myList)
    if length==0:
        return 0

    if length==1 and value == myList[0]:
        return value
    elif length==1:
        return -1

    if length>1:
        mid = length//2
        if value == myList[mid]:
            return value
        else:
            result = binarySearch(value, myList[:mid])
            if result!= -1:
                return result
            return binarySearch(value, myList[mid+1:])

binarySearch(number, list)
```
