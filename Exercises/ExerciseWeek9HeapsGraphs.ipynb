**Svar:** Heap sort har to faser.

**Fase 1: Byg en max-heap (bottom-up heapify)**

Listen tegnet som træ (læst niveau for niveau):
```
            55
          /    \
        22      34
       /  \    /  \
      10   2  99   68
```
Kør *sift down* på alle knuder med børn, fra den sidste og baglæns mod roden (34 → 22 → 55):

1. **34**: børn 99 og 68 → største er 99 → byt: `[55, 22, 99, 10, 2, 34, 68]`
2. **22**: børn 10 og 2 → 22 er størst → bliver
3. **55**: børn 22 og 99 → byt med 99; derefter børn 34 og 68 → byt med 68: `[99, 22, 68, 10, 2, 34, 55]`

Max-heap:
```
            99
          /    \
        22      68
       /  \    /  \
      10   2  34   55
```

**Fase 2: Sortér**

Hver runde: byt roden med det sidste element i heapen, flyt grænsen `|` én plads til venstre (elementet til højre er nu på sin endelige plads), og lad den nye rod synke ned (*sift down*) – kun inden for heapen.

| Runde | Byt | Sift down | Array efter runden |
|---|---|---|---|
| 1 | 99 ↔ 55 | 55 byttes med 68 (99 er sorteret og tæller ikke) | `[68, 22, 55, 10, 2, 34 \| 99]` |
| 2 | 68 ↔ 34 | 34 byttes med 55 | `[55, 22, 34, 10, 2 \| 68, 99]` |
| 3 | 55 ↔ 2 | 2 byttes med 34 | `[34, 22, 2, 10 \| 55, 68, 99]` |
| 4 | 34 ↔ 10 | 10 byttes med 22 | `[22, 10, 2 \| 34, 55, 68, 99]` |
| 5 | 22 ↔ 2 | 2 byttes med 10 | `[10, 2 \| 22, 34, 55, 68, 99]` |
| 6 | 10 ↔ 2 | – (kun ét element tilbage) | `[2 \| 10, 22, 34, 55, 68, 99]` |

**Resultat:** `[2, 10, 22, 34, 55, 68, 99]`

Tidskompleksitet: Fase 1 er O(n), og fase 2 er n runder á O(log n) → **O(n log n)** i alt. Sorteringen sker *in-place* (ingen ekstra liste).
