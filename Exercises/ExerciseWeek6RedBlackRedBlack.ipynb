*Exercise 1: Spot red black tree properties*

Which of the following are true properties of a red black tree

1. All nodes are either red or black - True (they cannot be other colours. But it can be a mix of the two).

2. Every leaf node must be red - False

3. The root and NIL nodes are black - True

4. You cannot have a red parent and a red child - True

5. Every path from the root to a leaf contains the same number of black nodes - True

6. All children of black nodes must be black - False

7. The tree must be a complete binary tree - False

8. The black height is the number of black nodes on any path from root to leaf - True

9. Every right child must be red -  False

10. A red child node can only have a red parent - False



A is invalid because you cannot have a red child and parent.

B is valid.

C is invalid because height of black is uneven. The black to the right is shorter than the others.
