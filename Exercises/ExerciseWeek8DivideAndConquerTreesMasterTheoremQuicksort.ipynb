Exercise 1: Identify divide and conquer algorithms

Identify which algorithms that are divide and conquer algorithms. For each, explain with your own words why/why not:

· Merge sort

· Insertion sort

· Binary search

· Linear search


*Merge sort*: Ja, det er en divide and conquer-algoritme.
Algoritmen opdeler listen rekursivt i to lige store halvdele (divide), sorterer hver halvdel rekursivt indtil del-listerne har en længde på 1 (conquer), og fletter derefter de sorterede del-lister sammen til ét samlet, sorteret array (combine).


*Insertion sort*: Nej, det er ikke en divide and conquer-algoritme.
Det er en iterativ/inkrementel algoritme. Den løser ikke problemet ved at splitte datasættet op i mindre uafhængige delproblemer, men opbygger derimod det sorterede array ét element ad gangen ved fortløbende at indsætte næste værdi på dens korrekte plads.



*Binary search*: Ja (med en lille teoretisk nuance).
I de fleste pensummer og lærebøger klassificeres den under divide and conquer, fordi den aktivt halverer søgerummet (divide) ved at sammenligne søgeværdien med midterelementet og derefter søger videre i den relevante halvdel (conquer). Formelt set kaldes den i datalogi ofte mere præcist for decrease and conquer, da den kun fortsætter ned i det ene delproblem og derfor ikke har et afsluttende trin, hvor delløsninger skal kombineres.



*Linear search*: Nej, det er ikke en divide and conquer-algoritme.
Algoritmen foretager en simpel sekventiel gennemgang af listen fra start til slut ét element ad gangen, indtil værdien findes eller listen er gennemløbet. Der sker ingen opdeling af problemet.

````python
import sympy as sp

# Definer variable og funktion
n = sp.symbols('n', integer=True, positive=True)
T = sp.Function('T')

# ---------------------------------------------------------
# Ligning 4: T(n) = T(n - 1) + 3
# Da dette er en standard differensligning med subtraktion,
# kan SymPy rsolve den direkte med en basisbetingelse T(1) = c.
# ---------------------------------------------------------
eq4 = sp.Eq(T(n), T(n - 1) + 3)
sol4 = sp.rsolve(eq4, T(n), {T(1): 1})
print(f"Ligning 4 lukket formel: T(n) = {sol4}")
# Resultat: 3*n - 2  => O(n)


# ---------------------------------------------------------
# Ligninger med division (T(n/b)):
# Standard rsolve håndterer T(n - 1), så vi laver en variabel-
# substitution n = b^k <=> k = log_b(n), hvor S(k) = T(b^k).
# ---------------------------------------------------------
k = sp.symbols('k', integer=True, positive=True)
S = sp.Function('S')

def solve_divide_and_conquer(a, b, f_k):
    """
    Løser T(n) = a * T(n/b) + f(n) ved at substituere n = b^k:
    S(k) = a * S(k - 1) + f(b^k)
    """
    eq = sp.Eq(S(k), a * S(k - 1) + f_k)
    # Løses med en vilkårlig basissag S(0) = 1 (svarende til T(1) = 1)
    sol_k = sp.rsolve(eq, S(k), {S(0): 1})

    # Substituer k tilbage til log(n, b)
    sol_n = sol_k.subs(k, sp.log(n, b))
    return sp.simplify(sol_n)

# 1. T(n) = 4 * T(n/4) + n  =>  f(n) = n = 4^k
sol1 = solve_divide_and_conquer(a=4, b=4, f_k=4**k)
print(f"Ligning 1 lukket formel: T(n) = {sol1}")
# Udtrykket svarer til: n*(log(n)/log(4) + 1) => O(n * log(n))

# 2. T(n) = 4 * T(n/4) + 2  =>  f(n) = 2
sol2 = solve_divide_and_conquer(a=4, b=4, f_k=2)
print(f"Ligning 2 lukket formel: T(n) = {sol2}")
# Udtrykket svarer til: 5*n/3 - 2/3 => O(n)

# 3. T(n) = 3 * T(n/9) + 1  =>  f(n) = 1, b = 9
sol3 = solve_divide_and_conquer(a=3, b=9, f_k=1)
print(f"Ligning 3 lukket formel: T(n) = {sol3}")
# Udtrykket svarer til: 3*sqrt(n)/2 - 1/2 => O(sqrt(n))

# 5. T(n) = T(n/3) + 2  =>  a = 1, b = 3, f(n) = 2
sol5 = solve_divide_and_conquer(a=1, b=3, f_k=2)
print(f"Ligning 5 lukket formel: T(n) = {sol5}")
# Udtrykket svarer til: 2*log(n)/log(3) + 1 => O(log(n))
````

### 1. $T(n) = 4 \cdot T\left(\frac{n}{4}\right) + n$

* **Find værdierne:**
  * $a = 4$ (4 delproblemer)
  * $b = 4$ (inputstørrelsen deles med 4)
  * $f(n) = n$ (arbejdet i toppen vokser lineært)
* **Regn bunden ud ($n^{\log_b a}$):**
  * $\log_b a = \log_4 4 = 1$
  * Bundfunktion: $n^1 = n$
* **Sammenlign bunden og toppen:**
  * Bund: $n$
  * Top $f(n)$: $n$
  * De vokser lige hurtigt (Tilfælde 2).
* **Konklusion:**
  $$T(n) = \Theta(n \log n)$$

### 2. $T(n) = 4 \cdot T\left(\frac{n}{4}\right) + 2$

* **Find værdierne:**
  * $a = 4$ (4 delproblemer)
  * $b = 4$ (inputstørrelsen deles med 4)
  * $f(n) = 2 = \Theta(1) = \Theta(n^0)$ (konstant arbejde)
* **Regn bunden ud ($n^{\log_b a}$):**
  * $\log_b a = \log_4 4 = 1$
  * Bundfunktion: $n^1 = n$
* **Sammenlign bunden og toppen:**
  * Bund: $n$
  * Top $f(n)$: $2$
  * Bunden $n$ vokser polynomielt hurtigere end konstanten $2$ (Tilfælde 1).
* **Konklusion:**
  $$T(n) = \Theta(n)$$

---

### 3. $T(n) = 3 \cdot T\left(\frac{n}{9}\right) + 1$

* **Find værdierne:**
  * $a = 3$ (3 delproblemer)
  * $b = 9$ (inputstørrelsen deles med 9)
  * $f(n) = 1 = \Theta(1) = \Theta(n^0)$ (konstant arbejde)
* **Regn bunden ud ($n^{\log_b a}$):**
  * $\log_b a = \log_9 3 = \frac{1}{2}$ (fordi $9^{1/2} = \sqrt{9} = 3$)
  * Bundfunktion: $n^{1/2} = \sqrt{n}$
* **Sammenlign bunden og toppen:**
  * Bund: $\sqrt{n}$
  * Top $f(n)$: $1$
  * Bunden $\sqrt{n}$ vokser polynomielt hurtigere end konstanten $1$ (Tilfælde 1).
* **Konklusion:**
  $$T(n) = \Theta(\sqrt{n})$$

---

### 4. $T(n) = T(n - 1) + 3$

* **Tjek om formen passer:**
  * Delproblemet reduceres med subtraktion ($- 1$) og ikke division ($\frac{n}{b}$).
  * **Master Theorem kan ikke anvendes.**
* **Løsning ved udfoldning (iteration):**
  * Hvert trin koster $3$ operationer og reducerer $n$ med $1$.
  * Fra $n$ ned til $0$ udføres trinnet $n$ gange: $3 + 3 + \dots + 3 = 3n$.
* **Konklusion:**
  $$T(n) = \Theta(n)$$

---

### 5. $T(n) = T\left(\frac{n}{3}\right) + 2$

* **Find værdierne:**
  * $a = 1$ (1 delproblem, da der står $1 \cdot T$)
  * $b = 3$ (inputstørrelsen deles med 3)
  * $f(n) = 2 = \Theta(1) = \Theta(n^0)$ (konstant arbejde)
* **Regn bunden ud ($n^{\log_b a}$):**
  * $\log_b a = \log_3 1 = 0$ (fordi $3^0 = 1$)
  * Bundfunktion: $n^0 = 1$
* **Sammenlign bunden og toppen:**
  * Bund: $1$
  * Top $f(n)$: $2$
  * Begge funktioner er konstanter og vokser i samme tempo (Tilfælde 2).
* **Konklusion:**
  Vi ganger med $\log n$:
  $$T(n) = \Theta(1 \cdot \log n) = \Theta(\log n)$$

Exercise 3: Divide and conquer

Implement a divide and conquer algorithm to find the maximum element of an unsorted list of integers.

Analyze the time complexity of your solution.

````python
def find_max(arr, low=0, high=None):
    """
    Finder det største element i arr[low..high] med divide and conquer.
    Vi arbejder med indekser i stedet for slicing, så der ikke kopieres lister.
    """
    if high is None:
        if not arr:
            raise ValueError("Listen er tom")
        high = len(arr) - 1

    # Basissag: ét element er sit eget maksimum
    if low == high:
        return arr[low]

    # Divide: del intervallet på midten
    mid = (low + high) // 2

    # Conquer: find maksimum i hver halvdel rekursivt
    left_max = find_max(arr, low, mid)
    right_max = find_max(arr, mid + 1, high)

    # Combine: det største af de to delresultater
    return left_max if left_max > right_max else right_max


# Tests
import random

assert find_max([7]) == 7
assert find_max([3, 9]) == 9
assert find_max([9, 3]) == 9
assert find_max([-5, -2, -8, -1]) == -1
assert find_max([4, 1, 4, 2]) == 4
assert find_max([12, 5, 33, 7, 1, 33, 0, 21]) == 33

for _ in range(1000):
    data = [random.randint(-1000, 1000) for _ in range(random.randint(1, 100))]
    assert find_max(data) == max(data)

print("Alle tests bestået")
print(find_max([12, 5, 33, 7, 1, 19, 0, 21]))
````

### Tidskompleksitet

* **Divide:** Beregning af `mid` tager konstant tid, $O(1)$.
* **Conquer:** To rekursive kald, hver på halvdelen af input: $2 \cdot T\left(\frac{n}{2}\right)$.
* **Combine:** Én sammenligning af `left_max` og `right_max`, $O(1)$.

Rekursionsligningen bliver derfor:
$$T(n) = 2 \cdot T\left(\frac{n}{2}\right) + 1, \quad T(1) = 1$$

**Master Theorem:**
* $a = 2$, $b = 2$, $f(n) = 1 = \Theta(n^0)$
* $\log_b a = \log_2 2 = 1$, så bundfunktionen er $n^1 = n$
* Bunden $n$ vokser polynomielt hurtigere end $f(n) = 1$ (Tilfælde 1)

$$T(n) = \Theta(n)$$

**Intuition:** Rekursionstræet har $n$ blade (ét pr. element) og $n - 1$ interne knuder, som hver laver præcis én sammenligning. Algoritmen laver altså i alt $n - 1$ sammenligninger, præcis som en simpel lineær gennemgang. Det giver mening, da ethvert element *skal* undersøges mindst én gang for at være sikker på maksimum, så $\Omega(n)$ er en nedre grænse, og vi kan ikke gøre det bedre.

**Pladskompleksitet:** Rekursionsdybden er $\lceil \log_2 n \rceil$, så kaldstakken bruger $O(\log n)$ ekstra plads (vi bruger indekser i stedet for slicing, så der kopieres ingen dellister).

Exercise 5: Trace divide and conquer

Given the list [8, 3, 4, 5, 2, 6, 9], show all the splits and merges that merge sort will result in on this array.

Listen har 7 elementer, så den kan ikke deles helt ligeligt. Vi bruger `mid = len(arr) // 2`, så venstre halvdel får det mindste antal elementer (med `mid = (len + 1) // 2` ville venstre side få det ekstra element i stedet, men princippet er det samme).

### Splits (divide)

```
                    [8, 3, 4, 5, 2, 6, 9]
                   /                     \
             [8, 3, 4]                 [5, 2, 6, 9]
             /       \                 /          \
          [8]      [3, 4]          [5, 2]        [6, 9]
                   /    \          /    \        /    \
                 [3]    [4]      [5]    [2]    [6]    [9]
```

1. `[8, 3, 4, 5, 2, 6, 9]` → `[8, 3, 4]` og `[5, 2, 6, 9]`
2. `[8, 3, 4]` → `[8]` og `[3, 4]`
3. `[3, 4]` → `[3]` og `[4]`
4. `[5, 2, 6, 9]` → `[5, 2]` og `[6, 9]`
5. `[5, 2]` → `[5]` og `[2]`
6. `[6, 9]` → `[6]` og `[9]`

### Merges (combine)

```
                 [3]    [4]      [5]    [2]    [6]    [9]
                   \    /          \    /        \    /
          [8]      [3, 4]          [2, 5]        [6, 9]
             \       /                 \          /
             [3, 4, 8]                 [2, 5, 6, 9]
                   \                     /
                    [2, 3, 4, 5, 6, 8, 9]
```

1. `[3]` + `[4]` → `[3, 4]`
2. `[8]` + `[3, 4]` → `[3, 4, 8]`
3. `[5]` + `[2]` → `[2, 5]`
4. `[6]` + `[9]` → `[6, 9]`
5. `[2, 5]` + `[6, 9]` → `[2, 5, 6, 9]`
6. `[3, 4, 8]` + `[2, 5, 6, 9]` → `[2, 3, 4, 5, 6, 8, 9]`

I alt giver det 6 splits og 6 merges ($n - 1 = 6$ for $n = 7$). Listerne ovenfor er grupperet efter type. I selve rekursionen er splits og merges flettet sammen: venstre halvdel (`[8, 3, 4]`) splittes og merges helt færdig, før højre halvdel påbegyndes. Koden nedenfor printer den faktiske rækkefølge.

````python
def merge_sort_trace(arr, depth=0):
    """Merge sort, der printer hvert split og merge med indrykning efter rekursionsdybde."""
    indent = "    " * depth
    if len(arr) <= 1:
        return arr

    # Divide
    mid = len(arr) // 2
    left, right = arr[:mid], arr[mid:]
    print(f"{indent}Split: {arr} -> {left} og {right}")

    # Conquer
    left = merge_sort_trace(left, depth + 1)
    right = merge_sort_trace(right, depth + 1)

    # Combine
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    print(f"{indent}Merge: {left} + {right} -> {merged}")
    return merged


result = merge_sort_trace([8, 3, 4, 5, 2, 6, 9])
assert result == [2, 3, 4, 5, 6, 8, 9]
print("\nResultat:", result)
````
