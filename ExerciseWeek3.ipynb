Discuss what the time complexity of the list operations are when
implemented with linked lists and with arrays.



| Operation | Array | Linked List |
| :--- | :--- | :--- |
| create    | O(1)  | O(1)        |
| size | O(1) | O(1) |
| isEmpty | O(1) | O(1) |
| get | O(1) | O(n) |
| insert | O(n) | O(n) |
|append | O(1) | O(1) |
|delete | O(n) | O(n) |
|search | O(n) | O(n) |
|iterate | O(n) | O(n) |
|concat | O(n1+n2) | O(n1+n2)  |
|sublist | O(k) | O(k) |
|clear | O(1) | O(1) |
