1. A good barista would make First in First out - Queue.
1. A bad barista would serve the Last-in, First-out - Stack

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head == None

    def last_element(self):
        return self.head.value if self.head else None

    def insert(self, value):
        new_node = Node(value)
        new_node.next = self.head
        self.head = new_node

lst = LinkedList()
print(lst.is_empty())
print(lst.last_element())

lst.insert(10)
lst.insert(22)
lst.insert(5)

print(lst.is_empty())
print(lst.last_element())
```

```python
def reverse_str(value):
        temp_stack = list(value)
        stack = []
        while temp_stack:
            stack.append(temp_stack.pop())
        return "".join(stack)

reverse_str("coffee")
```

[3,4,5,6]

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


class CustomerQueue:
    def __init__(self):
        self.head = None
        self.tail = None

    def is_empty(self):
        return self.head == None

    def insert(self, value):
        new_node = Node(value)
        if self.is_empty():
            self.head = new_node
        else:
            self.tail.next = new_node
        self.tail = new_node

    def next_customer(self):
        return self.head.value if self.head else None

    def remove_next(self):
        if self.is_empty():
            return None
        removed = self.head
        self.head = removed.next
        if self.head is None:
            self.tail = None
        return removed.value

    def length(self):
        i = 0
        current = self.head
        while current is not None:
            i += 1
            current = current.next
        return i


    def customers_ahead(self, value):
        i = 0
        current = self.head
        while current is not None:
            if current.value == value:
                return i
            i+=1
            current = current.next
        return None


queue = CustomerQueue()
queue.insert("Alice")
queue.insert("Bob")
queue.insert("Charlie")
print(queue.length())
print(queue.next_customer())
queue.length()

print(queue.customers_ahead("Charlie"))

queue.remove_next()

print(queue.next_customer())
queue.length()
```
